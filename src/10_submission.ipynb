# 10. Submission Generation
アンサンブル結果などを確認し、Kaggle等のコンペティション提出用の最終ファイルを作成・検証します。

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# パスの設定
PROJECT_ROOT = Path('/content/drive/MyDrive/jaggle_2026')
INPUT_DIR = PROJECT_ROOT / 'data' / 'input'
# 最新の日付（実行日）に合わせて変更してください
OUTPUT_DIR = PROJECT_ROOT / 'data' / 'output' / '20260806'
SUBMISSION_DIR = PROJECT_ROOT / 'data' / 'submission'
SUBMISSION_DIR.mkdir(parents=True, exist_ok=True)

sns.set_style('whitegrid')
```

## 1. 予測結果の読み込みと確認
ここでは、最もOOFスコアが良かった `Weighted Average` の推論結果をベースに最終提出ファイルを作成します。

```python
# 最良のモデル予測結果を読み込む
best_sub_file = OUTPUT_DIR / '20260806_08_ensemble_gbdt_advanced_features_weighted_avg.csv'

# ヘッダーなしで保存されているため、namesを指定して読み込む
sub_df = pd.read_csv(best_sub_file, names=['社員ID', '10年定着ラベル'])

display(sub_df.head())
print(f'Shape: {sub_df.shape}')
```

## 2. 予測値の分布確認
異常な予測値（すべて0や1に偏っていないか等）がないか、ヒストグラムで確認します。

```python
plt.figure(figsize=(10, 5))
sns.histplot(sub_df['10年定着ラベル'], bins=50, kde=True)
plt.title('Distribution of Predictions (Weighted Average)')
plt.xlabel('Predicted Probability')
plt.ylabel('Count')
plt.show()

print('--- 予測値の統計情報 ---')
print(sub_df['10年定着ラベル'].describe())
```

## 3. 最終提出ファイルの保存
フォーマット（ヘッダーの有無など）をコンペティションの要件に合わせて最終保存します。今回はヘッダー無しで保存します。

```python
# 提出用ファイルのパス
final_submission_path = SUBMISSION_DIR / 'final_submission.csv'

# 要件に合わせて保存 (ヘッダーなしの場合は header=False)
sub_df.to_csv(final_submission_path, index=False, header=False)
print(f'最終提出ファイルを保存しました: {final_submission_path}')
```
